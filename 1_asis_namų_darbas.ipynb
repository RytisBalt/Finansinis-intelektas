```python
#Pirmieji namų darbai
#1
import pandas as pd
import numpy as np
import datetime
import matplotlib.pyplot as plt

dates = pd.date_range('20190214', periods=6)
numbers = np.matrix( [[ 101, 103], [105.5, 75], [102, 80.3], [100, 85], [110, 98], [109.6, 125.7 ]] )
df = pd.DataFrame(numbers, index=dates, columns=['A','B'])
df
```

```python

date_range = pd.date_range(start='2000-01-01', end='2015-12-31', freq='D')
n = len(date_range)

delta = np.random.normal(0, 1, n)
delta[0] = 0
prices = np.cumsum(delta) + 150
plt.figure(figsize=(14, 8))
plt.plot(date_range, prices, color='b', alpha=0.7)
plt.title('Rinkos generavimas')
plt.xlabel('Metai')
plt.ylabel('Kaina')
plt.grid(True)
plt.show()
```

```python


data1 = pd.read_csv('/content/NFLX_1min_sample.csv', delimiter=',')

data1['timestamp'] = pd.to_datetime(data1['timestamp'])

start_date = pd.to_datetime('2024-02-14')
end_date = pd.to_datetime('2024-02-17')
filtered_data = data1[(data1['timestamp'].dt.date >= start_date.date()) &
                      (data1['timestamp'].dt.date <= end_date.date())]


filtered_data_6th_to_12th_hour = filtered_data[(filtered_data['timestamp'].dt.hour >= 6) &
                                                 (filtered_data['timestamp'].dt.hour < 12)]


plt.figure(figsize=(12, 6))
for day in filtered_data_6th_to_12th_hour['timestamp'].dt.date.unique():
    day_data = filtered_data_6th_to_12th_hour[filtered_data_6th_to_12th_hour['timestamp'].dt.date == day]
    plt.scatter(day_data['timestamp'], day_data['close'], label=str(day))

plt.title('Netflix stock Data between 6th and 12th Hour\nFebruary 14th - February 17th, 2024')
plt.xlabel('Time')
plt.ylabel('Price')
plt.xticks(rotation=45)
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

data1 = pd.read_csv('/content/NFLX_1min_sample.csv', delimiter=',')
data1['timestamp'] = pd.to_datetime(data1['timestamp'])

start_date = pd.to_datetime('2024-02-14')
end_date = pd.to_datetime('2024-02-17')
filtered_data = data1[(data1['timestamp'].dt.date >= start_date.date()) &
                      (data1['timestamp'].dt.date <= end_date.date())]



plt.figure(figsize=(12, 6))
for day in filtered_data['timestamp'].dt.date.unique():
    day_data = filtered_data[filtered_data['timestamp'].dt.date == day]
    plt.scatter(day_data['timestamp'], day_data['close'], label=str(day))

plt.title('Netflix stock Data between \nFebruary 14th - February 17th, 2024')
plt.xlabel('Time')
plt.ylabel('Price')
plt.xticks(rotation=45)
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.xlim(pd.Timestamp('2024-02-14 06:00:00'), pd.Timestamp('2024-02-17 12:00:00'))  # limit x-axis to desired date range
plt.show()
```

```python

data1 = pd.read_csv('/content/NFLX_1min_sample.csv', delimiter=',')


data1['timestamp'] = pd.to_datetime(data1['timestamp'])


data_16th_day = data1[data1['timestamp'].dt.day == 16]

time_diff = data_16th_day['timestamp'].diff().dt.total_seconds()

top_10_diff = time_diff.sort_values(ascending=False).head(10)

top_10_diff_minutes = top_10_diff / 60

print("10 didžiausių pertraukų: ")
for index, diff_minutes in top_10_diff_minutes.iteritems():
    start_time = data_16th_day.loc[index, 'timestamp'] - pd.Timedelta(seconds=time_diff.loc[index])
    print("Pertraukos pradžia:", start_time, "- Pertraukos pabaiga:", data_16th_day.loc[index, 'timestamp'], "- Pertraukos trukmė:", diff_minutes, "minučių")

```
